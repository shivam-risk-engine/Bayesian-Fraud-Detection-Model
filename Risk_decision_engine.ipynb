```python
P_fraud = float(input("Enter P(Fraud): "))
P_genuine = 1 - P_fraud

P_S_given_fraud = float(input("Enter P(S | Fraud): "))
P_S_given_genuine = float(input("Enter P(S | Genuine): "))

fraud_loss = float(input("Enter Fraud Loss: "))
genuine_loss = float(input("Enter Genuine Loss: "))

investigation_cost = float(input("Enter Investigation Cost: "))
fraud_save_percent = float(input("Enter % Fraud Loss Saved (e.g. 0.8 for 80%): "))


# STEP 1: BAYES

numerator = P_S_given_fraud * P_fraud
denominator = numerator + (P_S_given_genuine * P_genuine)

P_fraud_given_S = numerator / denominator


# STEP 2: EV (APPROVE)

EV_approve = (P_fraud_given_S * fraud_loss) + ((1 - P_fraud_given_S) * genuine_loss)


# STEP 3: EV (INVESTIGATE)

fraud_loss_after_save = fraud_loss * (1 - fraud_save_percent)

EV_investigate = (P_fraud_given_S * fraud_loss_after_save) + ((1 - P_fraud_given_S) * (genuine_loss + investigation_cost))



print("------ RESULTS -------")
print("P(Fraud | S): ", P_fraud_given_S)
print("EV Approve:", EV_approve)
print("EV Investigate:", EV_investigate)

# DECISION
if EV_approve < EV_investigate:
    print("Decision : APPROVE")
else:
    print("Decision : INVESTIGATE")


```
